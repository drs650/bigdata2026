# 실습 2

```python
f = open('guest.txt', 'a', encoding='utf-8')
population = 0

while True:
    name = input('참석하는 학생의 이름을 입력하세요 : ')
    if not name:
        break
    f.write(name + '\n') # \n : 엔터
    population += 1
f.close()
print(f'welcome')
print(f'총 인원수는 {population}명 입니다.')
```
