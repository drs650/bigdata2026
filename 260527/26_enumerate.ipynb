# 리스트의 for문

```python
pandas = ['러바오', '아이바오', '푸바오', '루이바오', '후이바오']
```

```python
for p in pandas:
    print(p) # 리스트의 값(요소)이 나온다.
```

```python
for i in range(5):
    print(i)
```

# enumerate 함수

for 인데스번호, 값 in enumerate(컬렉션자료형):

```python
for p in enumerate(pandas):
    print(p)
```
